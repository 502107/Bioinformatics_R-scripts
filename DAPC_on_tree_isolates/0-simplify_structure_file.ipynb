```python
import pandas as pd
```

```python
structure = pd.read_table('2-7clusters_seed8/Structure_syn_all_data_biallelic_cluster7.txt')
structure = structure.rename(columns={'Unnamed: 0': ''})
structure.to_csv('2-7clusters_seed8/structure7.txt', sep='\t', index=False)
```

```python
num_to_base = {-9: '?', 1: 'A', 2: 'T', 3: 'G', 4: 'C'}
for col_one, col_two in zip(structure.columns[1::2], structure.columns[2::2]):
    accession = col_one[:-2]
    structure[col_one] = [num_to_base[v] for v in structure[col_one]]
    structure[col_two] = [num_to_base[v] for v in structure[col_two]]
    structure[accession] = structure[col_one] + structure[col_two]
```

```python
keep_cols = [col for col in structure.columns if '_' not in col]
```
